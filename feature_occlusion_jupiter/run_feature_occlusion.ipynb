# Run Feature Occlusion

```python
import xarray as xr

from pkg.feature_occlusion.forecast_with_mask import run_multiple_forecasts_with_mask
```

```python
input_path = "../data/prediction_zeynep_22/input_data/graphcast_ready_input_zeynep_22_graphcast_small.nc"
input = xr.open_dataset(input_path)

climatology_path = "../data/data_for_climatology/era5_1979-2015_0218_climatology.nc"
climatology = xr.open_dataset(climatology_path)
```

```python
surface_variables = [
    "2m_temperature",
    ("10m_u_component_of_wind",
    "10m_v_component_of_wind"),
    "total_precipitation_6hr",
    "mean_sea_level_pressure",
]

atmospheric_variables = [
    "temperature",
    ("u_component_of_wind",
    "v_component_of_wind"),
    "geopotential",
    "vertical_velocity",
    "specific_humidity",
]

lower_troposphere = (850, 925, 1000)
mid_troposphere = (500, 600, 700)
upper_troposphere = (250, 300, 400)
stratospheric_levels  = (50, 100, 150, 200)

pressure_level_groups = [
    lower_troposphere,
    mid_troposphere,
    upper_troposphere,
    stratospheric_levels,
]

pressure_levels = [
    1000, 925, 850, 700, 600, 500, 400, 300, 250, 200, 150, 100, 50
]

# regions = [
#     "atlantic_jet_stream",
#     "storm_core_10utc",
#     "storm_core_16utc",
#     ("storm_core_10utc", "storm_core_16utc"),
#     "dynamic_uljs",
#     "storm_region",
#     "high_pressure_south",
#     "low_pressure_north",
#     "north_germany"
# ]

regions = [
    "atlantic_jet_stream",
    "storm_core",
    "dynamic_uljs",
    "storm_region",
    "high_pressure_south",
    "low_pressure_north",
    "north_germany"
]
```
